```python
import os
from google.colab import userdata
from groq import Groq

os.environ["Practice"] = userdata.get("Practice")
client = Groq()
```

```python
def summarize_emails(email_bodies):
  summary_list = []
  for number, email in enumerate(email_bodies):
    response = client.chat.completions.create(model="llama-3.3-70b-versatile",
         messages=[
             {
                  "role": "user",
                  "content": f"""I am going to send you the body of an email. I want you to summarize it in one line,
                  indicating the email's intent. The email is as follows: {email}"""
             }
             ]
             )

    text = response.choices[0].message.content
    print(f"E-Mail No. {number + 1}: {text} ")
    summary_list.append(f"E-Mail No. {number + 1}: {text} ")

  return summary_list
```

```python
summarize_emails(email_bodies)
```

```python
summary_list = summarize_emails(email_bodies)
```

This code block (or creates) a file and then writes each summary from your list onto a new line within that file.

```python
with open("summary_list.txt", "w", encoding="utf-8") as file:
  for summary in summary_list:
    file.write(summary + "\n")
```
